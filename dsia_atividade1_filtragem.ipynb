**Tema:** Filtragem

```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
url1 = "https://raw.githubusercontent.com/pietrapaz/dsia_dados/refs/heads/main/datasets/movies.csv"
url2 = "https://raw.githubusercontent.com/pietrapaz/dsia_dados/refs/heads/main/datasets/ratings.csv"

movies = pd.read_csv(url1)
ratings = pd.read_csv(url2)
```

```python
movies.head()
```

```python
ratings.head()
```

### Best-seller

```python
# agrupando as avaliacoes pelo id do filme
best_sellers = (ratings.groupby("movieId").size().reset_index(name="qtd").sort_values("qtd", ascending=False))
```

```python
# juntando best sellers com a tabela de filmes
best_sellers = best_sellers.merge(movies[["movieId", "title"]], on="movieId")
```

```python
# colunas que serao visualizadas
best_sellers = best_sellers[["movieId", "title", "qtd"]]
```

### Similaridade

```python
# avaliou o filme = 1; nao avaliou = 0
# criacao da tabela
matriz = (
    ratings.assign(interacao=1)
    .pivot_table(
        index="movieId",
        columns="userId",
        values="interacao",
        fill_value=0
    )
)
```

```python
similaridade = cosine_similarity(matriz)
```
