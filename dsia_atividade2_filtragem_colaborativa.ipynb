```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
url1 = "https://raw.githubusercontent.com/pietrapaz/dsia_dados/refs/heads/main/datasets/movies.csv"
url2 = "https://raw.githubusercontent.com/pietrapaz/dsia_dados/refs/heads/main/datasets/ratings.csv"

movies = pd.read_csv(url1)
ratings = pd.read_csv(url2)
```

```python
movies.head()
```

```python
ratings.head()
```

#### Best-seller

```python
# media das notas de cada filme
melhores_avaliados = (
    ratings.groupby("movieId")["rating"]
    .mean()
    .reset_index(name="media_rating")
    .sort_values("media_rating", ascending=False)
)
```

```python
# adicionando titulo do filme
melhores_avaliados = melhores_avaliados.merge(
    movies[["movieId", "title"]],
    on="movieId"
)
```

```python
# organizando as colunas
melhores_avaliados = melhores_avaliados[
    ["movieId", "title", "media_rating"]
]
```

```python
print(melhores_avaliados.head(10))
```

#### Similaridade

```python
# criacao da tabela
matriz = ratings.pivot_table(
    index="movieId",
    columns="userId",
    values="rating",
    fill_value=0
)
```

```python
matriz.head()
```

```python
similaridade = cosine_similarity(matriz)
```
